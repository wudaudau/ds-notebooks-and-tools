1) ImmPort — Immunology Database and Analysis Portal

🔗 A massive repository of **immunological data** from hundreds of clinical and mechanistic studies, including cytokine measurements and other biomarker data.  ￼

Why it’s useful
- Many studies include cytokine panels, flow cytometry, ELISA measurements and clinical metadata.
- You can filter for studies that have human subjects and clinical symptom scores.
- You’ll often find cytokine levels (e.g., IL-6, TNF-α, IL-10, etc.) alongside demographic and clinical variables that you can adapt to practice associations.

How to use
1.	Go to [ImmPort Shared Data](https://www.immport.org/shared/home) (no login required for many studies) and use the study search feature.   ￼
2.	Use keywords like “cytokine”, “ELISA”, “clinical symptoms”.
3.	Download the study’s CSV or tabular clinical datasets.
4.	You can merge “cytokine assay results” with clinical phenotypes for correlation practice. (2018 Nature https://www.nature.com/articles/sdata201815?utm_source=chatgpt.com)

Note: ImmPort houses hundreds of immunology datasets, but you’ll need to inspect study metadata to find cytokine + clinical severity variables that match PANSS-like scenarios.

2) dbGaP — NIH Database of Genotypes and Phenotypes

🔗 A large NIH archive of human “phenotype + genotype/biomarker” studies. (https://dbgap.ncbi.nlm.nih.gov/home/) ￼

Why it’s useful
	•	Includes clinical studies with cytokines, inflammatory markers (like CRP, IL-6) plus detailed phenotypes.
	•	Some portions of dbGaP data have open-access phenotype files (while others require controlled access).

Examples include:
	•	Studies of inflammation, psychiatric phenotypes, or immune biomarkers with structured clinical scores.
	•	Large cohorts like ARIC (Atherosclerosis Risk in Communities) and others that measure CRP and other biomarkers that you can correlate with clinical phenotypes.  ￼

How to use
	1.	Use the dbGaP search tool to filter by terms like:
	•	cytokine, inflammatory markers, CRP, IL-6
	2.	Download open-access phenotype tables if available.
	3.	Practice correlation and regression with biomarkers and standardized measures.

dbGaP often involves controlled content, but many phenotype tables (especially aggregate biomarker data) are publicly available.  ￼

3) Secondary Resources & Tools to Locate Immune/Clinical Data

🔎 PhenX Toolkit (https://en.wikipedia.org/wiki/PhenX_Toolkit?utm_source=chatgpt.com)

The PhenX Toolkit catalogues standardized measures used across many NIH studies.  ￼
	•	While not a dataset itself, it helps you identify which biomarker/clinical measures have common data elements across multiple public datasets.
	•	Useful for harmonizing variables (e.g., if you want a depression or psychosis score paired with cytokines for meta-analysis).

Tips For Discovery and Practice

🪪 Data Discovery Portal (https://arxiv.org/abs/2509.13524?utm_source=chatgpt.com)
	•	Use the NIAID Data Ecosystem Discovery Portal — a unified search engine across infectious and immune datasets.  ￼
→ Allows metadata search to find datasets combining biomarkers + clinical measures.

📁 Look for Gene Expression + Clinical Data

Many repositories like GEO (Gene Expression Omnibus) or ArrayExpress host studies with cytokine gene expression or protein arrays with clinical symptom annotations.
Even though these are not always serum cytokine levels, they can be used for exploratory practice (e.g., creating latent scores, correlation with symptom scales).
