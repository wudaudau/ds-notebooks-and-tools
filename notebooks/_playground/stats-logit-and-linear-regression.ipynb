Use Titanic dataset to explore the relationship between sex and survival rates.

```python
from pathlib import Path
import sys # for setting the importing env  

project_dir = Path.cwd().parent.parent # Adjust here to get the project_dir

# Add project_dir to sys.path at the first position
if project_dir not in sys.path:
    sys.path.insert(0, str(project_dir))
```

```python
import pandas as pd

from statsmodels.formula.api import logit, ols
```

```python
# Load data
titanic_data = pd.read_csv(project_dir/'data/titanic.csv')

print(titanic_data.shape)
print(titanic_data.head())
```

```python
# Create a logistic regression model and a linear regression model
logit_model = logit('Survived ~ C(Sex)', data=titanic_data).fit()
ols_model = ols('Survived ~ C(Sex)', data=titanic_data).fit()

# Print the summary of both models
print("Logistic Regression Model Summary:")
print(logit_model.summary())
print("\nLinear Regression Model Summary:")
print(ols_model.summary())
```

```python
# Create a logistic regression model and a linear regression model
logit_model = logit('Survived ~ C(Pclass)', data=titanic_data).fit()
ols_model = ols('Survived ~ C(Pclass)', data=titanic_data).fit()

# Print the summary of both models
print("Logistic Regression Model Summary:")
print(logit_model.summary())
print("\nLinear Regression Model Summary:")
print(ols_model.summary())
```
